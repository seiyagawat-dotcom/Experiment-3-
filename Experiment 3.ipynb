### Setup

```python
import pandas as pd

# Load the cars.csv dataset into a DataFrame named cars
cars = pd.read_csv('cars.csv')
```

### Problem A: Positional and Label-Based Slicing 

#### a. Shape and Column Names

```python
# Display the shape (rows, columns) and list of column names
print("Shape of DataFrame:", cars.shape)
print("\nColumn Names:\n", cars.columns.tolist())
```

#### b. Positional Slicing for Rows 6 through 10

```python
# Select rows 6 through 10 using positional slicing (iloc)
# Note: Since row 1 is index 0, rows 6 to 10 correspond to zero-based indices 5 to 9 (slice 5:10)
cars_6_to_10 = cars.iloc[5:10]
cars_6_to_10
```

#### c. Column Selection Using Column Labels

```python
# Display only Model, mpg, cyl, hp, and gear using column labels
cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
```

### Problem B: Model Lookup

#### a. Toyota Corolla Complete Row

```python
# Boolean indexing to retrieve the full row for Toyota Corolla
toyota = cars[cars['Model'] == 'Toyota Corolla']
toyota
```

#### b. Pontiac Firebird Subset

```python
# Boolean indexing with label-based column selection for Pontiac Firebird
pontiac = cars.loc[cars['Model'] == 'Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
pontiac
```

### Problem C: Multi-Model Subsetting

```python
# List of target vehicle models
target_models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']

# Select target models and retain specified columns
selected_cars = cars.loc[cars['Model'].isin(target_models), ['Model', 'mpg', 'cyl', 'hp', 'gear']]

# Display the subsetted DataFrame
print("Selected Cars DataFrame:")
display(selected_cars)

# Required check: Display shape (must be exactly 3 rows and 5 columns)
print("\nShape of selected_cars:", selected_cars.shape)
```
